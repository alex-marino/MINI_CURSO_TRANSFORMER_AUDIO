```python
import torch
import torchaudio
import librosa
import numpy as np
from transformers import Wav2Vec2Processor, Wav2Vec2FeatureExtractor, Wav2Vec2Model, WavLMModel, HubertModel


# Verifica se há uma GPU disponível
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Usando dispositivo: {device}")
```

```python
# Modelos disponíveis
MODELS = {
    "wavlm": "microsoft/wavlm-large",
    "wav2vec2": "facebook/wav2vec2-large-960h",
    "hubert": "facebook/hubert-large-ls960-ft"
}

# Carregar os processadores
processors = {
    "wavlm": Wav2Vec2FeatureExtractor.from_pretrained(MODELS["wavlm"]),
    "wav2vec2": Wav2Vec2Processor.from_pretrained(MODELS["wav2vec2"]),
    "hubert": Wav2Vec2Processor.from_pretrained(MODELS["hubert"])
}

# Carregar os modelos e mover para GPU
models = {
    "wavlm": WavLMModel.from_pretrained(MODELS["wavlm"]).to(device),
    "wav2vec2": Wav2Vec2Model.from_pretrained(MODELS["wav2vec2"]).to(device),
    "hubert": HubertModel.from_pretrained(MODELS["hubert"]).to(device)
}

# Colocar os modelos em modo de avaliação (evita atualizações de gradientes)
for model in models.values():
    model.eval()
```
