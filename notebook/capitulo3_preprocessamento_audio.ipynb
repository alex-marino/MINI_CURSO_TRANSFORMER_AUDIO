## 3.1 Padronização

```
import librosa, numpy as np

# Carregue um arquivo de áudio existente
y, sr = librosa.load("voz.wav", sr=16000)  # ajuste o caminho conforme necessário

# Normalização de amplitude
y = y / (np.max(np.abs(y)) + 1e-9)

# Segmentação em janelas fixas (4s)
segment_length = int(sr * 4)
chunks = [y[i:i+segment_length] for i in range(0, len(y), segment_length)]

# Padding do último segmento
if len(chunks[-1]) < segment_length:
    chunks[-1] = np.pad(chunks[-1], (0, segment_length - len(chunks[-1])))
len(chunks), sr
```

## 3.2 Visualização com `librosa`

```
import librosa.display, matplotlib.pyplot as plt

plt.figure(figsize=(10,3))
librosa.display.waveshow(y, sr=sr)
plt.title("Forma de onda")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.tight_layout()
plt.show()
```

## 3.3 Espectrogramas e transformadas

```
import matplotlib.pyplot as plt

S = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128)
S_db = librosa.power_to_db(S, ref=np.max)

plt.figure(figsize=(10,3))
librosa.display.specshow(S_db, y_axis='mel', x_axis='time', sr=sr)
plt.title("Mel-Espectrograma")
plt.colorbar(format="%+2.0f dB")
plt.tight_layout()
plt.show()
```
