## 6.2 Detecção de Patologias Vocais – Métricas

```
from sklearn.metrics import roc_auc_score, f1_score, brier_score_loss
import numpy as np

# Exemplo com dados simulados
y_true = np.r_[np.zeros(100), np.ones(100)]
y_pred = np.r_[np.random.uniform(0,0.4,100), np.random.uniform(0.6,1.0,100)]

print("AUC:", roc_auc_score(y_true, y_pred))
print("F1:", f1_score(y_true, y_pred > 0.5))
print("Brier:", brier_score_loss(y_true, y_pred))
```

## 6.3 Detecção de Deepfakes – SHAP para interpretação

```
import shap, numpy as np, xgboost as xgb
# Dados sintéticos: substitua por X_emb real
X_emb = np.random.normal(0, 1, (200, 64))
y = np.r_[np.zeros(100, dtype=int), np.ones(100, dtype=int)]

clf = xgb.XGBClassifier(n_estimators=200, max_depth=4, learning_rate=0.05, subsample=0.9, colsample_bytree=0.9, random_state=42)
clf.fit(X_emb, y)

explainer = shap.TreeExplainer(clf)
shap_values = explainer(X_emb)

# Visualização interativa pode exigir JS (em ambiente local)
shap.plots.beeswarm(shap_values, max_display=15)
```
